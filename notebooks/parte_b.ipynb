# PUNTO 5: Visualización 3D con Centroide

## 5.1 Importar librerías necesarias

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

print("✅ Librerías importadas correctamente")
```

## 5.2 Cargar datos 3D

```python
# Cargar puntos 3D desde puntos_3d.npy
puntos_3d = np.load('../puntos_3d.npy')

print(f"📊 Datos 3D cargados:")
print(f"   Forma: {puntos_3d.shape}")
print(f"   Total de puntos: {len(puntos_3d)}")
print(f"\n   Primeros 5 puntos:")
print(puntos_3d[:5])
```

## 5.3 Calcular centroide

```python
# Calcular el centroide (promedio de todos los puntos)
centroide = np.mean(puntos_3d, axis=0)

print(f"\n🎯 Centroide calculado:")
print(f"   Coordenadas: [{centroide[0]:.4f}, {centroide[1]:.4f}, {centroide[2]:.4f}]")
print(f"\n   X promedio: {centroide[0]:.4f}")
print(f"   Y promedio: {centroide[1]:.4f}")
print(f"   Z promedio: {centroide[2]:.4f}")
```

## 5.4 Visualización 3D

```python
# Crear visualización 3D
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Graficar puntos 3D (en rojo)
ax.scatter(puntos_3d[:, 0], puntos_3d[:, 1], puntos_3d[:, 2], 
           c='red', marker='o', s=50, alpha=0.6, label='Puntos 3D')

# Graficar centroide (en azul)
ax.scatter(centroide[0], centroide[1], centroide[2], 
           c='blue', marker='*', s=500, edgecolors='black', linewidths=2,
           label='Centroide', zorder=5)

# Configurar etiquetas y título
ax.set_xlabel('X', fontsize=12, fontweight='bold')
ax.set_ylabel('Y', fontsize=12, fontweight='bold')
ax.set_zlabel('Z', fontsize=12, fontweight='bold')
ax.set_title('Visualización 3D: Puntos y Centroide', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)

# Mejorar visibilidad
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n✅ Visualización 3D generada")
print(f"   Centroide marcado en AZUL: {centroide}")
```

# PUNTO 6: KMeans en el Dataset Iris

## 6.1 Importar librerías

```python
from sklearn.datasets import load_iris
from src.clustering.kmeans_model import KMeansModel
from src.utils.visualization import Visualizer

print("✅ Librerías para Iris importadas correctamente")
```

## 6.2 Cargar dataset Iris

```python
# Cargar dataset Iris
iris = load_iris()
X_iris = iris.data
y_iris = iris.target

print(f"📊 Dataset Iris cargado:")
print(f"   Forma de X: {X_iris.shape}")
print(f"   Clases: {iris.target_names}")
print(f"   Features: {iris.feature_names}")
print(f"\n   Primeras 5 muestras:")
print(X_iris[:5])
print(f"\n   Primeras 5 etiquetas: {y_iris[:5]}")
```

## 6.3 Aplicar KMeans con k=3

```python
# Entrenar KMeans con 3 clusters
kmeans_iris = KMeansModel(n_clusters=3, random_state=42)
kmeans_iris.fit(X_iris)

centroids_iris = kmeans_iris.get_centroids()
predicted_iris = kmeans_iris.get_labels()

print("✅ KMeans entrenado en Iris")
print(f"\nInformación del modelo:")
info = kmeans_iris.get_info()
for key, value in info.items():
    print(f"  {key}: {value}")

print(f"\nCentroides:")
for i, c in enumerate(centroids_iris):
    print(f"  Cluster {i}: {c}")
```

## 6.4 Visualización de clusters (2D proyección)

```python
# Visualizar en 2D usando las primeras 2 features (sepal length vs sepal width)
X_iris_2d = X_iris[:, :2]
centroids_iris_2d = centroids_iris[:, :2]

plt.figure(figsize=(10, 7))
scatter = plt.scatter(X_iris_2d[:, 0], X_iris_2d[:, 1], 
                      c=predicted_iris, cmap='viridis', s=80, alpha=0.6, edgecolors='black')
plt.scatter(centroids_iris_2d[:, 0], centroids_iris_2d[:, 1], 
            c='red', marker='*', s=400, edgecolors='black', linewidths=2, 
            label='Centroides')

plt.xlabel(iris.feature_names[0], fontsize=12)
plt.ylabel(iris.feature_names[1], fontsize=12)
plt.title('KMeans en Iris Dataset (Proyección 2D)', fontsize=14, fontweight='bold')
plt.colorbar(scatter, label='Cluster')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n✅ Visualización de clusters Iris generada")
```

## 6.5 Análisis de resultados

```python
print(f"\n{'='*70}")
print(f"ANÁLISIS DE RESULTADOS")
print(f"{'='*70}")

# Distribución de clusters
unique_clusters = np.unique(predicted_iris)
print(f"\n📊 Distribución de clusters:")
for cluster_id in unique_clusters:
    count = np.sum(predicted_iris == cluster_id)
    pct = (count / len(predicted_iris)) * 100
    print(f"   Cluster {cluster_id}: {count} muestras ({pct:.1f}%)")

# Comparación con etiquetas reales
from sklearn.metrics import adjusted_rand_score, silhouette_score

ari = adjusted_rand_score(y_iris, predicted_iris)
silhouette = silhouette_score(X_iris, predicted_iris)

print(f"\n📈 Métricas de evaluación:")
print(f"   Adjusted Rand Index (ARI): {ari:.4f}")
print(f"   Silhouette Score: {silhouette:.4f}")

print(f"\n💡 Interpretación:")
print(f"   - ARI = {ari:.4f}: ", end="")
if ari > 0.7:
    print("Excelente concordancia con clases reales")
elif ari > 0.5:
    print("Buena concordancia con clases reales")
else:
    print("Concordancia moderada con clases reales")
    
print(f"   - Silhouette = {silhouette:.4f}: ", end="")
if silhouette > 0.5:
    print("Clusters bien definidos")
elif silhouette > 0.3:
    print("Clusters razonablemente definidos")
else:
    print("Clusters con solapamiento")
```

# PUNTO 7: Filtro de Kalman

## 7.1 Importar librerías

```python
from src.filters.kalman_filter import KalmanFilter

print("✅ Librerías para Filtro de Kalman importadas")
```

## 7.2 Generar datos de ejemplo (posición con ruido)

```python
# Generar trayectoria real según especificación: pos = 0.1*(3*t - t**2)
np.random.seed(42)
n_samples = 50
tiempo = np.linspace(0, 3, n_samples)

# Trayectoria real: pos = 0.1*(3*t - t²)
trayectoria_real = 0.1 * (3 * tiempo - tiempo**2)

# Mediciones con ruido
ruido = np.random.normal(0, 0.05, n_samples)
mediciones = trayectoria_real + ruido

print(f"📊 Datos generados:")
print(f"   Muestras: {n_samples}")
print(f"   Tiempo: 0 a 3 segundos")
print(f"   Trayectoria: pos = 0.1*(3*t - t²)")
print(f"   Ruido: Normal(μ=0, σ=0.05)")
print(f"\n   Primeras 5 mediciones: {mediciones[:5]}")
```

## 7.3 Aplicar Filtro de Kalman

```python
# Crear y aplicar el filtro de Kalman
kalman = KalmanFilter(process_variance=1e-5, measurement_variance=0.0025)
estimaciones_completas = kalman.filter(mediciones)

# Eliminar el primer valor (estado inicial) para que coincida con mediciones
estimaciones = estimaciones_completas[1:]

print("✅ Filtro de Kalman aplicado")
print(f"\nParámetros del filtro:")
print(f"   Varianza de proceso (Q): {1e-5}")
print(f"   Varianza de medición (R): {0.0025}")
print(f"\n   Primeras 5 estimaciones: {estimaciones[:5]}")
```

## 7.4 Visualización de resultados

```python
# Visualizar trayectoria real, mediciones y estimaciones
plt.figure(figsize=(12, 6))
plt.plot(tiempo, trayectoria_real, 'g-', linewidth=2, label='Trayectoria Real (pos = 0.1*(3t - t²))', zorder=3)
plt.scatter(tiempo, mediciones, c='red', s=30, alpha=0.5, label='Mediciones (con ruido)')
plt.plot(tiempo, estimaciones, 'b-', linewidth=2, label='Estimación Kalman', zorder=2)

plt.xlabel('Tiempo (segundos)', fontsize=12)
plt.ylabel('Posición', fontsize=12)
plt.title('Filtro de Kalman: Estimación de Trayectoria Cuadrática', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n✅ Visualización del Filtro de Kalman generada")
```

## 7.5 Análisis de error

```python
# Calcular errores
error_mediciones = np.abs(mediciones - trayectoria_real)
error_kalman = np.abs(estimaciones - trayectoria_real)

mae_mediciones = np.mean(error_mediciones)
mae_kalman = np.mean(error_kalman)

print(f"\n{'='*70}")
print(f"ANÁLISIS DE ERROR")
print(f"{'='*70}")
print(f"\nError Absoluto Medio (MAE):")
print(f"   Mediciones con ruido: {mae_mediciones:.6f}")
print(f"   Estimación Kalman:    {mae_kalman:.6f}")
print(f"\n📉 Reducción de error: {((mae_mediciones - mae_kalman) / mae_mediciones * 100):.2f}%")

print(f"\n💡 Conclusión:")
print(f"   El Filtro de Kalman suaviza la trayectoria cuadrática pos = 0.1*(3t - t²),")
print(f"   reduciendo el ruido y mejorando la estimación en {((mae_mediciones - mae_kalman) / mae_mediciones * 100):.1f}%.")
```

# PUNTO 8: Filtro Complementario

## 8.1 Importar librerías

```python
from src.filters.complementary_filter import ComplementaryFilter

print("✅ Librerías para Filtro Complementario importadas")
```

## 8.2 Generar datos de dos sensores

```python
# Generar señal real
np.random.seed(42)
n_samples = 100
tiempo = np.linspace(0, 10, n_samples)
señal_real = np.sin(tiempo)

# Sensor 1: Alta frecuencia, bajo drift (ej: giroscopio)
sensor1 = señal_real + np.random.normal(0, 0.1, n_samples)

# Sensor 2: Baja frecuencia, con drift (ej: acelerómetro)
drift = 0.02 * tiempo
sensor2 = señal_real + drift + np.random.normal(0, 0.05, n_samples)

print(f"📊 Datos de sensores generados:")
print(f"   Muestras: {n_samples}")
print(f"   Señal real: sin(t)")
print(f"   Sensor 1: Alta frecuencia (σ=0.1)")
print(f"   Sensor 2: Con drift (0.02t + σ=0.05)")
```

## 8.3 Probar diferentes valores de alpha

```python
# Probar con diferentes valores de alpha
alphas = [0.2, 0.5, 0.8]
resultados = {}

print(f"\n{'='*70}")
print(f"PROBANDO DIFERENTES VALORES DE ALPHA")
print(f"{'='*70}")

for alpha in alphas:
    comp_filter = ComplementaryFilter(alpha=alpha)
    estimacion = comp_filter.filter_series(sensor1, sensor2)
    
    # Calcular error
    error = np.mean(np.abs(estimacion - señal_real))
    resultados[alpha] = {'estimacion': estimacion, 'error': error}
    
    print(f"\nAlpha = {alpha}:")
    print(f"   Error medio: {error:.6f}")

# Encontrar el mejor alpha
mejor_alpha = min(resultados.keys(), key=lambda k: resultados[k]['error'])
print(f"\n✅ Mejor alpha: {mejor_alpha} (error = {resultados[mejor_alpha]['error']:.6f})")
```

## 8.4 Visualización de resultados

```python
# Visualizar señales
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Subplot 1: Señal real vs sensores
ax1 = axes[0, 0]
ax1.plot(tiempo, señal_real, 'g-', linewidth=2, label='Señal Real', zorder=3)
ax1.plot(tiempo, sensor1, 'r-', alpha=0.5, label='Sensor 1 (ruidoso)')
ax1.plot(tiempo, sensor2, 'b-', alpha=0.5, label='Sensor 2 (con drift)')
ax1.set_xlabel('Tiempo')
ax1.set_ylabel('Valor')
ax1.set_title('Señales de Entrada', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Subplots 2-4: Resultados con diferentes alphas
for idx, alpha in enumerate(alphas):
    row = (idx + 1) // 2
    col = (idx + 1) % 2
    ax = axes[row, col]
    
    estimacion = resultados[alpha]['estimacion']
    error = resultados[alpha]['error']
    
    ax.plot(tiempo, señal_real, 'g-', linewidth=2, label='Señal Real', zorder=3)
    ax.plot(tiempo, estimacion, 'm-', linewidth=2, label=f'Filtro α={alpha}', zorder=2)
    ax.set_xlabel('Tiempo')
    ax.set_ylabel('Valor')
    ax.set_title(f'Filtro Complementario (α={alpha}, Error={error:.4f})', fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    if alpha == mejor_alpha:
        ax.set_facecolor('#f0fff0')

plt.tight_layout()
plt.show()

print("\n✅ Visualización del Filtro Complementario generada")
```

## 8.5 Análisis de alpha óptimo

```python
print(f"\n{'='*70}")
print(f"ANÁLISIS DE ALPHA")
print(f"{'='*70}")

print(f"\n📊 Resumen de errores:")
for alpha in sorted(resultados.keys()):
    error = resultados[alpha]['error']
    es_mejor = " ⭐ MEJOR" if alpha == mejor_alpha else ""
    print(f"   α = {alpha}: Error = {error:.6f}{es_mejor}")

print(f"\n💡 Interpretación de alpha:")
print(f"   • α cercano a 0: Mayor peso al Sensor 2 (menos drift)")
print(f"   • α cercano a 1: Mayor peso al Sensor 1 (menos ruido)")
print(f"   • α = {mejor_alpha}: Balance óptimo para estos datos")

print(f"\n✅ Conclusión:")
print(f"   El Filtro Complementario con α={mejor_alpha} produce")
print(f"   la mejor estimación, combinando las ventajas de ambos sensores.")
```
