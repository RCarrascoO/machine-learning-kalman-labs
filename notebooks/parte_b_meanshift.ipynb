## 1. Importar librerías

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import DataLoader, print_data_info
from src.clustering.meanshift_model import MeanShiftModel
from src.utils.visualization import Visualizer

print("✅ Librerías importadas correctamente")
```

## 2. Cargar datos de MeanShift

```python
# Cargar datos de entrenamiento
X, labels = DataLoader.load_meanshift_data()

print(f"Datos cargados:")
print(f"  Forma de X: {X.shape}")
print(f"  Forma de labels: {labels.shape}")
print(f"  Primeras 5 filas de X:\n{X[:5]}")
print(f"  Primeras 5 labels: {labels[:5]}")
```

## 3. Información sobre los datos

```python
print_data_info(X, labels, title="INFORMACIÓN DE DATOS - MeanShift")
```

## 4. Gráfico de datos originales (Figura A)

```python
# Graficar datos originales con etiquetas reales
Visualizer.plot_2d_clusters(X, labels, title="Figura A: Datos Originales MeanShift")
plt.show()
```

## 5. Entrenar modelo MeanShift

```python
# Crear y entrenar el modelo
meanshift = MeanShiftModel(bandwidth='scott')
meanshift.fit(X)

print("✅ Modelo MeanShift entrenado")
print(f"Información del modelo:")
info = meanshift.get_info()
for key, value in info.items():
    print(f"  {key}: {value}")
```

## 6. Obtener centroides y etiquetas

```python
# Obtener resultados
centroids = meanshift.get_centroids()
predicted_labels = meanshift.get_labels()

print(f"Centroides del modelo:")
print(centroids)
print(f"\nForma de centroides: {centroids.shape}")
print(f"Etiquetas predichas (primeras 10): {predicted_labels[:10]}")
```

## 7. Gráfico de clusters predichos (Figura B)

```python
# Graficar clusters con centroides
Visualizer.plot_2d_clusters(X, predicted_labels, centroids, 
                            title="Figura B: Clusters MeanShift (sin centroides resaltados)")
plt.show()
```

## 8. Gráfico comparativo (Figura C)

```python
# Graficar comparación original vs clusters
Visualizer.plot_training_vs_clusters(X, labels, X, predicted_labels, centroids)
plt.show()
```

## 9. Información de etiquetas y datos

```python
# Información sobre etiquetas
unique_clusters = np.unique(predicted_labels)
print(f"\n{'='*50}")
print(f"INFORMACIÓN DE ETIQUETAS - MeanShift")
print(f"{'='*50}")
print(f"Clusters identificados: {unique_clusters}")
print(f"Total de clusters detectados: {len(unique_clusters)}")
print(f"\nDistribución por cluster:")

for i, cluster_id in enumerate(unique_clusters):
    count = np.sum(predicted_labels == cluster_id)
    percentage = (count / len(predicted_labels)) * 100
    centroid = centroids[i]
    print(f"  Cluster {int(cluster_id)}: {count} puntos ({percentage:.1f}%)")
    print(f"    Centroide: {centroid}")
```

## 10. Predicciones en puntos de test

```python
# Puntos de test según el enunciado
test_points = np.array([
    [-7, -6],
    [1.5, -6.5],
    [7.9, 0.5],
    [5.5, 10],
])

# Realizar predicciones
predictions = meanshift.predict(test_points)

print(f"\n{'='*50}")
print(f"PREDICCIONES EN PUNTOS DE TEST")
print(f"{'='*50}")
for i, (point, pred) in enumerate(zip(test_points, predictions)):
    print(f"Punto {i+1}: {point} → Cluster {int(pred)}")
```

## 11. Visualizar datos de test con predicciones

```python
# Graficar datos de entrenamiento + puntos de test
plt.figure(figsize=(10, 8))

# Datos de entrenamiento
scatter = plt.scatter(X[:, 0], X[:, 1], c=predicted_labels, 
                      cmap='viridis', s=50, alpha=0.6, edgecolors='k',
                      label='Datos de entrenamiento')

# Centroides
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', 
           marker='X', s=300, edgecolors='black', linewidths=2,
           label='Centroides')

# Puntos de test
plt.scatter(test_points[:, 0], test_points[:, 1], c='yellow', 
           marker='s', s=200, edgecolors='black', linewidths=2,
           label='Puntos de test')

plt.xlabel('Característica 1')
plt.ylabel('Característica 2')
plt.title('MeanShift: Datos de entrenamiento + Predicciones')
plt.colorbar(scatter, label='Cluster')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 12. Resumen de resultados

```python
print(f"\n{'='*60}")
print(f"RESUMEN DE RESULTADOS - MEANSHIFT")
print(f"{'='*60}")
print(f"✅ Datos cargados: {X.shape[0]} puntos, {X.shape[1]} dimensiones")
print(f"✅ Modelos entrenados: MeanShift")
print(f"✅ Clusters detectados automáticamente: {len(unique_clusters)}")
print(f"✅ Centroides generados: {centroids.shape[0]} centros")
print(f"✅ Predicciones realizadas: {len(test_points)} puntos de test")
print(f"\nClase de cada punto de test:")
for i, pred in enumerate(predictions, 1):
    print(f"  Punto {i}: Pertenece a la clase {int(pred)}")
print(f"\n{'='*60}")
```
